# AI-DeFi Risk Intelligence: EDA and Modeling
This notebook demonstrates the Exploratory Data Analysis, SMOTE class imbalance handling, and Model Comparison as required by the final project rubric.

## 1. Data Loading and EDA

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/datasets/blockchain_data.csv')
df.head()
```

```
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='is_fraud')
plt.title('Original Class Distribution (Highly Imbalanced)')
plt.show()
```

## 2. SMOTE and Preprocessing

```
X = df.drop(columns=['is_fraud'])
y = df['is_fraud']

from imblearn.over_sampling import SMOTE
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X, y)

plt.figure(figsize=(6,4))
sns.countplot(x=y_resampled)
plt.title('Class Distribution after SMOTE')
plt.show()
```

## 3. Model Comparison

```
results = pd.read_csv('../models/model_comparison.csv')
results.head()
```

```
plt.figure(figsize=(10,5))
sns.barplot(data=results, x='Model', y='F1-Score')
plt.title('Model F1-Score Comparison')
plt.show()
```

## 4. SHAP Explainability on XGBoost

```
import shap
import pickle

with open('../models/xgboost_classifier.pkl', 'rb') as f:
    xgb = pickle.load(f)
with open('../models/scaler.pkl', 'rb') as f:
    scaler = pickle.load(f)
with open('../data/datasets/feature_names.pkl', 'rb') as f:
    feature_names = pickle.load(f)

explainer = shap.TreeExplainer(xgb)
X_test = np.load('../data/datasets/X_test_scaled.npy')
shap_values = explainer.shap_values(X_test[:100])

shap.summary_plot(shap_values, X_test[:100], feature_names=feature_names)
```
